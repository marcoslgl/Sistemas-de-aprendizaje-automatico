# Parte 5 - Detectar outliers mediante IQR

## Ejercicio 5

Calcula:

$$
IQR=Q_3-Q_1
$$

y:

$$
L_{inferior}=Q_1-1.5IQR
$$

$$
L_{superior}=Q_3+1.5IQR
$$

Después localiza los clientes cuyo `gasto_mensual` esté fuera de estos límites y guarda el resultado en `outliers_gasto`.

```python
# TODO: calcula Q1, Q3 e IQR
Q1 = df['gasto_mensual'].quantile(0.25)
Q3 = df['gasto_mensual'].quantile(0.75)
IQR = Q3 - Q1

# TODO: calcula los límites inferior y superior
limite_inferior = Q1 - 1.5 * IQR
limite_superior = Q3 + 1.5 * IQR

# TODO: muestra los resultados
print(f"Q1: {Q1:.2f}")
print(f"Q3: {Q3:.2f}")
print(f"IQR: {IQR:.2f}")
print(f"Límite inferior: {limite_inferior:.2f}")
print(f"Límite superior: {limite_superior:.2f}")
```

```python
# TODO: crea outliers_gasto
outliers_gasto = df[
    (df['gasto_mensual'] < limite_inferior) | (df['gasto_mensual'] > limite_superior)
]

# TODO: muestra los posibles outliers
print(f"Posibles outliers: {len(outliers_gasto)} ({len(outliers_gasto) / len(df) * 100:.2f} %)")
outliers_gasto
```

Calcula frecuencias absolutas, relativas y porcentajes mediante `value_counts()`.

```python
# TODO: analiza dispositivo
def frecuencias(columna):
    return pd.DataFrame({
        'Absoluta': df[columna].value_counts(),
        'Relativa': df[columna].value_counts(normalize=True).round(4),
        'Porcentaje (%)': (df[columna].value_counts(normalize=True) * 100).round(2),
    })

frecuencias('dispositivo')
```

```python
# TODO: desarrolla aquí el análisis de la hipótesis seleccionada
# Hipótesis A: ¿los clientes con mayores ingresos gastan más mensualmente?

# 1. Selección de variables
hipotesis = df[['ingresos_anuales', 'gasto_mensual']].dropna()

# 2. Estadística descriptiva
print(hipotesis.describe().round(2))
print()

# 3. Visualización
plt.figure(figsize=(8, 5))
plt.scatter(hipotesis['ingresos_anuales'], hipotesis['gasto_mensual'], alpha=0.6, edgecolor='black')
plt.title('Ingresos anuales frente a gasto mensual')
plt.xlabel('Ingresos anuales (€)')
plt.ylabel('Gasto mensual (€)')
plt.show()

# 4. Correlación
print(f"r = {hipotesis['ingresos_anuales'].corr(hipotesis['gasto_mensual']):.3f}")
print()

# Gasto medio por tramos de ingresos (cuartiles)
hipotesis['tramo_ingresos'] = pd.qcut(hipotesis['ingresos_anuales'], 4,
                                      labels=['Bajo', 'Medio-bajo', 'Medio-alto', 'Alto'])
print(hipotesis.groupby('tramo_ingresos', observed=True)['gasto_mensual'].agg(['mean', 'median']).round(2))
```
